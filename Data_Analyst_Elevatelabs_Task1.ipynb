```python
#Calculate the percentage of missing values in each column
missing_percentage = df.isnull().sum()/len(df)*100
print(missing_percentage)

#You can also filter to see only columns with missing values
columns_with_missing = missing_percentage[missing_percentage > 0]
print(columns_with_missing)
```

```python
#Create a heatmap of missing values
sns.heatmap(df.isnull(),cbar = False)
plt.show()
```

```python
#Calculate the number of missing values for each column
missing_values = df.isnull().sum()

#Plot the bar chart
plt.figure(figsize=(10,6))
missing_values.plot(kind='bar',color='skyblue')
plt.title('Number of missing values per column')
plt.xlabel('Columns')
plt.ylabel('Number of Missing values')
plt.show()
```

```python
#To plot the percentage instead of the count
missing_percentage = (df.isnull().sum()/len(df)) * 100
plt.figure(figsize=(10,6))
missing_percentage.plot(kind='bar',color='salmon')
plt.title('Percentage of missing Values per Column')
plt.xlabel('Columns')
plt.ylabel('Percentage')
plt.ylim(0,100)
plt.show()
```

```python
missing_values_per_row = df.isnull().sum(axis=1)
plt.figure(figsize=(10,6))
missing_values_per_row.plot(kind='bar',color='purple')
plt.title('Number of Missing Values Per row')
plt.xlabel('Row Index')
plt.ylabel('Number of Missing Values')
plt.show()
```

```python
#Standardize a 'Country' column by converting to lowercase and stripping whitespaces
df['title'] = df['title'].str.lower().str.strip()
df
```

```python
duplicates = df.duplicated(keep=False)

#count the number of duplicated rows
num_duplicates = duplicates.sum()
print(f"Total number of duplicated rows: {num_duplicates}")

#you can also use a simple sum to count only the subsequent duplicates
print(f"Number of rows that will be dropped: {df.duplicated().sum()}")
```

```python
#Remove all duplicate rows
df_no_duplicates = df.drop_duplicates()

#You can also remove duplicates based on a subset of columns, for example:
df_unique_customers = df.drop_duplicates(subset=['title'])

df
```
